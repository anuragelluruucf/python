## Q1. Movie Ticket Counter

A cinema charges ticket prices based on age:
  - Below 5 years   : Free
  - 5 to 17 years   : Rs. 100
  - 18 to 59 years  : Rs. 200
  - 60 and above    : Rs. 120

Ask the customer for their age and print the ticket price.

Sample run:
  Enter your age: 65
  Your ticket price is Rs. 120

**How the solution works:** Check the age bands in order. After age < 5 fails, the next branch already knows the age is at least 5. Use <= to include 17 and 59.

```python
age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age")
elif age < 5:
    print("Your ticket is free")
elif age <= 17:
    print("Your ticket price is Rs. 100")
elif age <= 59:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")
```

## Q2. Electricity Bill Calculator

The electricity board charges as follows:
  - First 100 units       : Rs. 3 per unit
  - Next 100 units (101-200): Rs. 5 per unit
  - Above 200 units       : Rs. 8 per unit

Take the number of units consumed and print the total bill.

Sample run:
  Enter units consumed: 250
  Total bill: Rs. 1200
  (100*3 + 100*5 + 50*8 = 300 + 500 + 400)

**How the solution works:** This is slab pricing: only the units above each threshold get the higher rate. For 250 units: 300 + 500 + 400 = Rs. 1200.

```python
units = float(input("Enter units consumed: "))

if units < 0:
    print("Invalid units")
else:
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = 100 * 3 + (units - 100) * 5
    else:
        bill = 100 * 3 + 100 * 5 + (units - 200) * 8

    print("Total bill: Rs.", bill)
```

## Q3. Multiplication Table for a Kid

A little kid wants to learn tables. Ask for a number and print
its multiplication table from 1 to 10 using a for loop.

Sample run:
  Enter a number: 7
  7 x 1 = 7
  7 x 2 = 14
  ...
  7 x 10 = 70

**How the solution works:** range(1, 11) produces 1 through 10. The stop value 11 is excluded.

```python
number = int(input("Enter a number: "))

for multiplier in range(1, 11):
    print(number, "x", multiplier, "=", number * multiplier)
```

## Q4. Savings Piggy Bank

Riya puts money into her piggy bank every day. Her goal is
Rs. 1000. Using a while loop, keep asking how much she saved
today and add it to the total. Stop when the total reaches or
crosses Rs. 1000, then print how many days it took.

Sample run:
  Enter today's savings: 300
  Enter today's savings: 450
  Enter today's savings: 400
  Goal reached in 3 days! Total saved: Rs. 1150

**How the solution works:** Start with zero savings and zero days. Add each valid daily amount, then count that day. A zero-savings day still counts. The loop stops when total is at least 1000. Negative entries are rejected and do not count as days.

```python
total = 0
days = 0

while total < 1000:
    savings = float(input("Enter today's savings: "))
    if savings < 0:
        print("Enter zero or a positive amount")
    else:
        total = total + savings
        days = days + 1

print("Goal reached in", days, "days! Total saved: Rs.", total)
```

## Q5. ATM PIN Check

An ATM's correct PIN is 4321. The user gets only 3 attempts.
  - If the PIN is correct, print "Access granted".
  - If wrong, print "Wrong PIN" and ask again.
  - After 3 wrong attempts, print "Card blocked".

Sample run:
  Enter PIN: 1111
  Wrong PIN
  Enter PIN: 4321
  Access granted

**How the solution works:** Count every attempt. A correct PIN ends the loop immediately with break. The final check blocks the card only if no attempt succeeded. The PIN is text because it is an identifier, not a quantity.

```python
attempts = 0
access_granted = False

while attempts < 3:
    pin = input("Enter PIN: ")
    attempts = attempts + 1

    if pin == "4321":
        print("Access granted")
        access_granted = True
        break
    else:
        print("Wrong PIN")

if access_granted == False:
    print("Card blocked")
```

## Q6. Even and Odd Counter

A teacher wants to know how many roll numbers from 1 to N are
even and how many are odd. Ask for N and use a for loop to
count them.

Sample run:
  Enter N: 15
  Even numbers: 7
  Odd numbers: 8

**How the solution works:** % gives the remainder. A remainder of zero after division by 2 means even. range(1, n + 1) includes n.

```python
n = int(input("Enter N: "))
even_count = 0
odd_count = 0

if n < 1:
    print("Enter a positive integer")
else:
    for number in range(1, n + 1):
        if number % 2 == 0:
            even_count = even_count + 1
        else:
            odd_count = odd_count + 1

    print("Even numbers:", even_count)
    print("Odd numbers:", odd_count)
```

## Q7. Number Guessing Game

The secret number is 27 (store it in a variable). Keep asking
the player to guess until they get it right. After each wrong
guess, tell them "Too high" or "Too low". At the end, print
how many guesses they took.

Sample run:
  Guess the number: 50
  Too high
  Guess the number: 20
  Too low
  Guess the number: 27
  Correct! You took 3 guesses.

**How the solution works:** != means not equal. Keep looping until the guess equals 27. Increment the counter before checking so the successful guess also counts.

```python
secret = 27
guesses = 0
guess = 0

while guess != secret:
    guess = int(input("Guess the number: "))
    guesses = guesses + 1

    if guess > secret:
        print("Too high")
    elif guess < secret:
        print("Too low")
    else:
        print("Correct! You took", guesses, "guesses.")
```

## Q8. Exam Result with Grade

Take marks (out of 100) of a student and print the grade:
  - 90 and above : A
  - 75 to 89     : B
  - 50 to 74     : C
  - 35 to 49     : D
  - Below 35     : Fail
If the marks entered are less than 0 or more than 100, print
"Invalid marks".

**How the solution works:** Validate the marks first. Then check thresholds from highest to lowest. Only the first matching branch runs.

```python
marks = float(input("Enter marks out of 100: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 35:
    print("Grade: D")
else:
    print("Fail")
```

## Q9. Sum of Digits on a Bill Number

A shop gives a free gift if the sum of the digits of the bill
number is more than 20. Take the bill number (as an integer)
and use a while loop with % and // to find the sum of its
digits. Print whether the customer gets a gift.

Sample run:
  Enter bill number: 98765
  Sum of digits: 35
  Congratulations! You win a free gift.

**How the solution works:** For 987: 987 % 10 gives the last digit, 7; 987 // 10 removes that digit, leaving 98. Repeat until no digits remain. A sum of exactly 20 does not qualify. Bill number 0 has digit sum 0.

```python
bill_number = int(input("Enter bill number: "))

if bill_number < 0:
    print("Enter a non-negative bill number")
else:
    number = bill_number
    digit_sum = 0

    while number > 0:
        digit = number % 10
        digit_sum = digit_sum + digit
        number = number // 10

    print("Sum of digits:", digit_sum)
    if digit_sum > 20:
        print("Congratulations! You win a free gift.")
    else:
        print("No free gift.")
```

## Q10. Staircase Pattern

A game developer needs a simple staircase made of '*'.
Ask for the number of steps and print the pattern using
nested for loops.

Sample run:
  Enter number of steps: 4
  *
  * *
  * * *
  * * * *

**How the solution works:** The outer loop chooses the row. The inner loop prints that many stars. end=" " keeps stars on the same line; the final print() moves to the next line.

```python
steps = int(input("Enter number of steps: "))

if steps < 1:
    print("Enter a positive number of steps")
else:
    for row in range(1, steps + 1):
        for column in range(row):
            print("*", end=" ")
        print()
```
